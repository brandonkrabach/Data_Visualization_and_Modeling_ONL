## The reproducible analysis workflow

1. **Question:** What do we want to learn?
2. **Data:** Which observations and features can help answer it?
3. **Evidence:** What should we calculate and display to answer the question?
4. **Check:** Does the result make sense in context? Do a few original records support the meaning of the calculation?
5. **Conclusion:** What claim is supported?
6. **Limitation:** What should we avoid concluding?

## Investigation 2

### Step 1: Question

State a specific question that your dataset can answer. Define any important groups or comparisons.


**What is your research question 2?**

Your answer: What percent of Pokémon weigh more than 100 kg? I will convert `weight` from hectograms to kilograms and count the Pokémon over 100 kg (not including exactly 100 kg), out of all 998 Pokémon.


### Prediction — before calculating


**What do you predict the result will be, and why?**

Your answer: I think around 12% of Pokémon will weigh more than 100 kg. Legendary Pokémon tend to be heavy, and quite a few non-legendary Pokémon are big too, like Metagross and Snorlax. Some small Pokémon are also very heavy, like Cosmoem, so size alone doesn't decide it.


### Step 2: Data


Reuse the DataFrame and initial inspection from Investigation 1. Identify the observations and features relevant to this question; inspect additional features only if needed.


**Which observations and features will you use? Explain relevant units, any filtering, and how you will handle missing or not-applicable values using the dictionary. For a percentage, define the denominator.**

Your answer: I will use all 998 Pokémon with no filtering, and the `name` and `weight` features. The dictionary says `weight` is in grams, but the values match hectograms (Bulbasaur's 69 = 6.9 kg), so I will divide by 10 to convert to kilograms. `weight` has no missing values, so every Pokémon is included. The denominator is all 998 Pokémon, and the numerator is the number of Pokémon weighing more than 100 kg. A Pokémon at exactly 100 kg does not count.


### Step 3: Evidence

Write the code that answers your question. Break the analysis into separate cells for individual operations, as in the lecture, and add cells as needed. Display intermediate results where useful and the final result that answers the question. Explain any new calculated features. You may display a grouped summary directly without assigning it to a variable.


```python
pokemon_df["weight_kg"] = pokemon_df["weight"] / 10
pokemon_df[["name", "weight", "weight_kg"]].head()
```

```python
pokemon_df["over_100_kg"] = (pokemon_df["weight_kg"] > 100).astype(int)
pokemon_df[["name", "weight_kg", "over_100_kg"]].head(10)
```

```python
over_100_kg_count = pokemon_df["over_100_kg"].sum()
over_100_kg_count
```

```python
over_100_kg_percent = over_100_kg_count / len(pokemon_df["over_100_kg"]) * 100
over_100_kg_percent
```

```python
print("Over 100 kg:", over_100_kg_count)
```

```python
print("All Pokémon:", len(pokemon_df["over_100_kg"]))
```

```python
print("Percent:", round(over_100_kg_percent, 2))
```

**Which specific value or comparison answers your question? Explain how your calculations connect the selected data to that result.**

Your answer: 163 out of 998 Pokémon weigh more than 100 kg, which is about 16.33%. I divided `weight` by 10 to convert hectograms to kilograms and saved it as `weight_kg`. Then I made an indicator, `over_100_kg`, that is 1 when `weight_kg` is greater than 100 and 0 otherwise. Summing the indicator gives the count of Pokémon over 100 kg (163), and dividing by all 998 Pokémon and multiplying by 100 gives the percentage.
